# 문제 셀

## 1. 열 선택

```python
# `fare` 열 하나를 선택해 타입을 출력하세요.
print(type(titanic["fare"]))
# `survived`, `pclass`, `sex`, `age` 4개 열을 선택해 타입과 `shape`를 출력하세요.
t_spsa = titanic[["survived", "pclass", "sex", "age"]]
print(type(t_spsa))
print(t_spsa.shape)
# 1번과 2번의 타입이 다른 이유를 주석 한 줄로 쓰세요.
# 주석 : titanic은 DataFrame 형식의 배열이 들어간 객체를 호출하는 변수이다. DataFrame은 Series들이 합쳐져 모여있는 2차원 구조를 띄고 있다. 만약 하나의 열만 호출하고자 하면 Series 형식으로 반환되지만, 여러 열을 호출하면 2차원 구조인 DataFrame 형식으로 반환된다. 
```

## 2. loc vs iloc

```python
# `loc`으로 인덱스 10~15 행의 `sex`, `age` 열을 선택해 출력하세요.
t_loc_sa = titanic.loc[10:15, ["sex", "age"]]
print(t_loc_sa)
# `iloc`으로 같은 열(`sex`는 2번째, `age`는 3번째 위치)을 행 위치 10:15로 선택해 출력하세요.
t_iloc_23 = titanic.iloc[10:15, [2, 3]]
print(t_iloc_23)
# 두 결과의 행 수를 각각 출력하고, 왜 다른지 주석으로 쓰세요.
print(t_loc_sa.shape[0])
print(t_iloc_23.shape[0])
# 주석: loc은 시작행과 끝행을 포함하여 행을 선택하지만, iloc은 인덱싱 슬라이싱 방법을 활용해 행을 선택한다. loc은 시작행인 10부터 끝행인 15까지 포함한 행을 선택하고, iloc은 시작행인 10은 포함하나 끝행인 15는 포함하지 않은 채 14까지의 행만 선택한다.
```

## 3. iloc 응용

```python
# iloc으로 마지막 3행의 앞 4개 열만 선택해 출력하세요. (힌트: 음수 위치)
print(titanic.iloc[-3 :  : 1, 0 : 4 : 1])
```

## 4. 조건 마스크

```python
# 나이가 10살 미만인지 검사하는 마스크를 만들어 `head(5)`를 출력하세요.
mask_10 = titanic["age"] < 10
print(mask_10.head(5))
# 마스크로 10살 미만 승객 수를 출력하세요.
print(titanic[mask_10].shape[0])
```

## 5. 복합 조건

```python
# 각각 조건을 만족하는 행 수를 출력하세요.

# 3등석 **그리고** 남성 **그리고** 생존한 승객 (`&`)
mask1 = (titanic["pclass"] == 3) & (titanic["sex"] == "male") & (titanic["survived"] == 1)
print(titanic[mask1].shape[0])
# 1등석 **또는** 요금 100 이상인 승객 (`|`)
mask2 = (titanic["pclass"] == 1) | (titanic["fare"] >= 100)
print(titanic[mask2].shape[0])
# 혼자 탑승하지 **않은** 승객 (`~`, `alone` 열 사용)
mask3 = ~titanic["alone"]
print(titanic[mask3].shape[0])
```

## 6. isin · between

```python
# `isin`으로 Cherbourg 또는 Queenstown에서 탑승한 승객을 골라 `shape`를 출력하세요.
print(titanic[titanic["embark_town"].isin(["Cherbourg", "Queenstown"])].shape)
# `between`으로 나이 20~30세(양 끝 포함) 승객을 골라 `shape`와 나이의 min·max를 출력하세요.
age_between = titanic[titanic["age"].between(20, 30)]
print(age_between.shape, age_between["age"].min(), age_between["age"].max())
# 같은 범위를 양 끝 미포함(`inclusive="neither"`)으로 바꿔 `shape`를 출력하세요.
print(titanic[titanic["age"].between(20, 30, inclusive = "neither")].shape)
```

## 7. loc[조건, 열]

```python
# 12살 이하 그리고 3등석 승객의 age, sex, survived 열만 kids 변수에 담고, 행 수와 생존률(survived 평균, 소수 둘째 자리)을 출력하세요.
kids_mask = (titanic["age"] <= 12) & (titanic["pclass"] == 3)
kids = titanic.loc[kids_mask, ["age", "sex", "survived"]]

print(kids.shape[0], f"{round(kids['survived'].mean() * 100, 2)}%")
```

## 8. query

```python
# 변수 `min_fare = 50`을 만들고, `query`로 Cherbourg 탑승 **그리고** 요금이 `min_fare` 이상인 승객의 `shape`를 출력하세요.
min_fare = 50
print(titanic.query("embark_town == 'Cherbourg' and fare >= @min_fare").shape)
# 5-1번 조건을 `query`로 다시 써서 행 수가 같은지 확인하세요.
print(titanic.query("pclass == 3 and sex == 'male' and survived == 1").shape[0] == titanic[mask1].shape[0])
```

## 9. 정렬

```python
# 1. `pclass` 오름차순, 같은 등급 안에서는 `fare` 내림차순으로 정렬해 `pclass`, `fare`, `sex`, `age` 열의 상위 3행을 출력하세요.
print(titanic.sort_values(by = ["pclass", "fare"], ascending = [True, False])[["pclass", "fare", "sex", "age"]].head(3))
# 2. 요금 내림차순으로 정렬한 결과를 변수에 담고, `iloc[0]`과 `loc[0]`의 `fare` 값을 각각 출력하세요. 두 값이 다른 이유를 주석으로 쓰세요.
titanic_f = titanic.sort_values(by="fare", ascending=False)
print(titanic_f.iloc[0]["fare"], titanic_f.loc[0]["fare"])
# 주석: iloc[0]은 정렬된 결과의 첫 번째 행 위치이고, loc[0]은 정렬 전의 인덱스 이름표가 0인 행이다.
```

## 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석

```python
# 질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

# 남성 · 나이 20~40세(`between`) · 1·2등석(`isin`) 조건으로 `pclass`, `age`, `fare`, `survived` 열을 선택하세요.
male_chose = titanic.loc[(titanic["sex"] == "male") & (titanic["age"].between(20, 40)) & (titanic["pclass"].isin([1, 2])), ["pclass", "age", "fare", "survived"]]
# 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
print(male_chose.shape[0])
print(f"{round(male_chose["survived"].mean() * 100, 2)}%")
print(male_chose.sort_values(by = "fare", ascending = False).head(3))
# 같은 조건을 `query`로 써서 행 수를 검증하세요.
print(titanic.query("sex == 'male' and age >= 20 and age <= 40 and pclass in [1, 2]").shape[0])
# 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.
print(f"{round(titanic.loc[(titanic["sex"] == "male") & (titanic["age"].between(20, 40)) & (titanic["pclass"] == 3), ["pclass", "age", "fare", "survived"]]["survived"].mean() * 100, 2)}%")
```

20~40세 남성 1·2등석의 생존률은 24.27%, 동일조건 3등석의 생존률은 15.09%이다.
이것으로 유전무죄 무전유죄 돈없으면 죽어야하는 시대가 바로 타이타닉 침몰 시기였다는 것을 알 수 있다.
